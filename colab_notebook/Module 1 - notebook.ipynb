### A1 — Putting a value into a message

A variable holds something. An f-string drops it into text.

```python
name = "Aisyah"
print(f"Hello, {name}!")
```

**Notice:** the `f` before the quote is what makes `{name}` get replaced.
Without it you would print the word `{name}` literally.

### A3 — Is this word inside that message?

This single line is the heart of the whole chatbot.

```python
msg = "where is my order"

print("order"  in msg)
print("parcel" in msg)
```

**Notice:** `in` gives you `True` or `False` — nothing else. The computer is not
reading the sentence. It is checking whether those exact letters appear
somewhere inside it.

`True` for `order`, `False` for `parcel` — even though a person would say both
messages mean the same thing.

### A4 — A trap: capital letters

Computers treat `H` and `h` as different characters.

```python
print("hello" in "Hello there")            # False
print("hello" in "Hello there".lower())    # True
```

**Notice:** the first line is `False`, which surprises most people. `.lower()`
makes everything small first, so the comparison works.

**Remember this one** — you will need it in Part B, and it explains something
odd that happens later.

### A5 — Choosing what to say

`if` / `elif` / `else` picks one reply out of several.

```python
msg = "where is my order"

if "hello" in msg:
    print("Bot: Hi! How can I help?")
elif "order" in msg:
    print("Bot: Your order is on the way.")
else:
    print("Bot: Sorry, I didn't catch that.")
```

**Notice:** Python checks the branches **from the top, and stops at the first one
that matches.** The rest are never looked at.

*Change `msg` to `"hello there"` and run again. Then try `"banana"`.*

### A6 — Doing it for several messages at once

A `for` loop runs the same code on each item in a list.

```python
messages = ["hello", "where is my order", "banana"]

for msg in messages:
    print("You:", msg)

    if "hello" in msg:
        print("Bot: Hi! How can I help?")
    elif "order" in msg:
        print("Bot: Your order is on the way.")
    else:
        print("Bot: Sorry, I didn't catch that.")

    print()
```

**Notice:** exactly the same `if`/`elif`/`else` as A5 — only now it runs three
times, once per message. You can see all three answers at once, which is handy
for testing.

### A7 — A real conversation

`while True` repeats forever, so the bot keeps talking until `break` stops it.

```python
while True:
    msg = input("You: ").lower()

    if "hello" in msg:
        print("Bot: Hi! How can I help?")
    elif "order" in msg:
        print("Bot: Your order is on the way.")
    elif "bye" in msg:
        print("Bot: Goodbye!")
        break
    else:
        print("Bot: Sorry, I didn't catch that.")
```

**Notice:** `break` is the only way out. Type `bye` to stop it properly.

**If it will not stop,** press the **■ stop button** where ▶ used to be. You will
need that button in every module.

## Stretch — ELIZA, 1966 *(only if you finish early)*

In 1966 a researcher at MIT built a program called ELIZA that imitated a
therapist. People talked to it for hours. Some refused to believe it was a
program even after it was explained to them.

It used the same technique you just used: matching words and printing a reply.
No learning of any kind.

### Then the questions

**Question 1:** Where do ELIZA's replies come from?

**Question 2 (the harder one):** Your bot uses the same technique and fools
nobody. What is ELIZA doing that yours is not?

<br><br><br><br><br><br><br><br><br><br>

### The answer

**1.** *(Probe 1.)* ELIZA's replies come almost entirely from *your own words*,
turned around. `my brother made me come here` → `Your brother made you come
here?` It swaps *my* for *your* and *me* for *you*, adds a question mark, and
hands your sentence back.

**2.** Two things, and the second one is the surprise.

**It only ever asks.** Your bot tries to **answer**, so when it fails it fails
visibly. ELIZA only ever asks a question back, so a user who wants to believe it
understands will do the work of believing.

**And its fallback is far better than yours.** *(Probes 2 and 3.)* Your bot has
one reply for everything it cannot match: *Sorry, I didn't catch that.* ELIZA had
two moves. It could give a content-free nudge — *Please go on*, *In what way?* —
which keeps the conversation moving without claiming anything. That is what
probe 2 should have shown you: it never once said *I don't understand*.

Its second move was to **reach back into what you said earlier** and resurface
it: *Earlier you said your sister…* That is what probe 3 was hunting for.

That second move is why people were convinced. A program that never looks stuck
reads as understanding, even when it understood nothing.

**If probe 3 found nothing**, that is a real result, not a failed experiment.
The version you talked to is a rebuild from 2005, and rebuilds do not all carry
every feature of the original. Which is a useful lesson in itself: *the copy is
not the thing.* Much of what people believe about ELIZA comes from playing with
simplified descendants rather than Weizenbaum's program.

### Why this is worth knowing

Handling "I don't know" gracefully is one of the hardest parts of building a
chatbot, and one of the most important. It has a name, and it is on your
glossary: **fallback**.

There is a live argument that modern systems went *backwards* on this. A 1966
program, when it had nothing, said so and asked you to keep talking. Some modern
systems, when they have nothing, produce a confident answer that is simply not
true. You will see this happen for yourself in Module 6.

One footnote: the ELIZA most people have played with online is not Weizenbaum's.
A simplified copy spread in the 1970s and most web versions descend from it —
which is why the real thing is usually more capable than people expect. The
original source code was only recovered and made to run again recently.

JavaScript implementation by N. Landsteiner, 2005.*
